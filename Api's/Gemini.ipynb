### 1. Basic Text Generation

```python
from google import genai

client=genai.Client(api_key=GEMINI_API_KEY)

interactions=client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="What is AI?"
    )
print(interactions.output_text)
```

```python
# exploring temperature 

interaction=client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Explain what do we need Activation function",
    generation_config={
        "temperature":0.3
    }
)
print(interaction.output_text)
```

### Streaming Text 

```python
stream=client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Generate me Sample email ,from Amazon to Piyush Thakur providing ppo of 78 lakhs per annum.",
    stream=True
)

for event in stream:
    if event.event_type=="step.delta":
        if event.delta.type=="text":
            print(event.delta.text, end="")
```
