```python
from dotenv import load_dotenv
import os
load_dotenv()
```

### 1. Basic Text Generation

```python
from google import genai

client=genai.Client(api_key=GEMINI_API_KEY)

interactions=client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="What is AI?"
    )
print(interactions.output_text)
```

```python
# exploring temperature 

interaction=client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Explain what do we need Activation function",
    generation_config={
        "temperature":0.3
    }
)
print(interaction.output_text)
```

### 2. Streaming Text 

```python
stream=client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Generate me Sample email ,from Amazon to Piyush Thakur providing ppo of 78 lakhs per annum.",
    stream=True
)

for event in stream:
    if event.event_type=="step.delta":
        if event.delta.type=="text":
            print(event.delta.text, end="")
```

### 3. System Instructions

```python
from google import genai

client=genai.Client(api_key=GEMINI_API_KEY)

interaction=client.interactions.create(
    model="gemini-3.1-flash-lite",
    system_instruction="""You are a Python Teacher ,
        explain in simple language ,
        and give small code examples""",
    input="What is decorator in Python?"
)

print(interaction.output_text)
```

```python
response=client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Give me 5 startups ideas",
    generation_config={
        "temperature":0.7
    }
)

print(response.output_text)
```

### 3. Multi-turn conversations / conversation memory

```python
response1=client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="My name is Piyush and I perform Cindy regularly" 
)

print(response1.id)
```

```python
response2=client.interactions.create(
    model="gemini-3.1-flash-lite",
    previous_interaction_id=response1.id,
    input="What do i do regularly?"
)
print(response2.output_text)
```

```python
response3=client.interactions.create(
    model="gemini-3.1-flash-lite",
    previous_interaction_id=response2.id,
    input="What is my name?"
)
print(response3.output_text)
```

### 4. Multimodal input

```python
from google import genai

image=client.files.upload(file="image.png")

response = client.interactions.create(
    model="gemini-3.1-flash-lite",
    input=[
        {
            "type": "image",
            "uri": image.uri,
            "mime_type": image.mime_type
        },
        {
            "type": "text",
            "text": "Describe this image in detail."
        }
    ]
)

print(response.output_text)
```
