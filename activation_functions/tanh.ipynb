### Tanh Activation Function -> maps the input to values between -1 and 1

### 1. Numpy Implementation

```python
import numpy as np 

def tanh(x):
    return (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))

x=np.array([-3,-2,-1,0,1,2,3])

print(tanh(x))
```

### 2. Pytorch Implementation

```python
import torch

x=torch.tensor([-3,-2,-1,0,1,2,3])

print(torch.tanh(x))
```

### 3. Visualization 

```python
import matplotlib.pyplot as plt 

x=np.linspace(-10,10,200)

plt.figure(figsize=(8,6))
plt.plot(x,tanh(x))
plt.title("Tanh Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()
         
```

### It cuts the x axis at Exactly zero - hence zero centred
